```python
import requests
import pandas as pd
import os
from sodapy import Socrata
from datetime import datetime, timedelta

# ------ Client to Initiate Connection -----------
client = Socrata("data.cityofnewyork.us", 
                 os.getenv("app_token"),
                 username = os.getenv("username"),
                 password = os.getenv("password")
                 )


# ------ Rodent Data Functions ------
# These functions return pandas dataframes of the number of 311 rodent incidents by ZIP, Address, Borough, 

def rodents_zip_date():
    """Returns a dataframe of 311 Rodent incidents by zip code"""
    six_months_ago = (datetime.now() - timedelta(days= 182)).strftime("%Y-%m-%dT%H:%M:%S")
    results = client.get(
        "erm2-nwe9",
        select="incident_zip, complaint_type, count(*) AS num_complaints",
        where=f"complaint_type='Rodent' AND created_date >= '{six_months_ago}'",
        group="incident_zip, complaint_type",
        limit=50000
    )
    results_df = pd.DataFrame(results)

    return(results_df)   
```
